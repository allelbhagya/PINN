```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt

folder = "../00_data/data"
files = sorted(glob.glob(os.path.join(folder, "*")))[:10]

for f in files:
    try:
        df = pd.read_csv(f)

        if "Voltage_measured" not in df.columns:
            continue

        print(df.head())

        df.plot(figsize=(10, 5))
        plt.title(os.path.basename(f))
        plt.tight_layout()
        plt.show()

    except Exception as e:
        pass
```
