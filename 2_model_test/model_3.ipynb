```python
import os
import glob
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

folder = "/Users/bhagya/Desktop/code/PINN/00_data/data/discharge_files/"

files = glob.glob(os.path.join(folder, "*.csv"))

dfs = []
for f in files:
    df = pd.read_csv(f)
    if set(["Voltage_measured", "Current_measured", "Temperature_measured", "Current_load", "Voltage_load", "Time"]).issubset(df.columns):
        dfs.append(df[["Voltage_measured", "Current_measured", "Temperature_measured", "Current_load", "Voltage_load", "Time"]])

data = pd.concat(dfs, ignore_index=True)

y = data["Voltage_measured"].values
X = data[["Current_measured", "Temperature_measured", "Current_load", "Voltage_load", "Time"]].values


X.shape, y.shape
```

```python

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_rnn = X_scaled.reshape(X_scaled.shape[0], X_scaled.shape[1], 1)

X_train, X_test, y_train, y_test = train_test_split(X_rnn, y, test_size=0.2, random_state=42)

model = Sequential()
model.add(SimpleRNN(32, input_shape=(X_train.shape[1], X_train.shape[2]), activation="tanh"))
model.add(Dense(1))

model.compile(optimizer="adam", loss="mse")
model.fit(X_train, y_train, epochs=20, batch_size=32, validation_data=(X_test, y_test))

y_pred = model.predict(X_test).flatten()

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
```
